```python
from xgboost import XGBClassifier
from ucimlrepo import fetch_ucirepo
from sklearn.metrics import (accuracy_score, precision_score,confusion_matrix, f1_score,recall_score, auc, roc_curve,ConfusionMatrixDisplay)
from sklearn.model_selection import train_test_split
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
from xgboost import plot_importance
import tracemalloc
import seaborn as sns
import matplotlib.patches as mpatches
import numpy as np
```

```python
phishing = fetch_ucirepo(id=327)
x = phishing.data.features
y = phishing.data.targets.values.ravel()

print(f"Dataset loaded: {x.shape[0]} samples, {x.shape[1]} features")
```

```python
plt.rcParams.update({'font.family': 'DejaVu Sans', 'figure.dpi': 130,
                     'axes.spines.top': False, 'axes.spines.right': False})
# ── PLOT 1: Class Distribution ───────────────────────────────
fig, ax = plt.subplots(figsize=(7, 4))

labels  = ['Legitimate (-1)', 'Suspicious (0)', 'Phishing (1)']
counts  = [(y == -1).sum(), (y == 0).sum(), (y == 1).sum()]
colors  = ['#2e7d32', '#f57f17', '#c62828']
bars    = ax.bar(labels, counts, color=colors, width=0.5,
                 edgecolor='white', linewidth=1.2)

for bar, val in zip(bars, counts):
    pct = val / len(y) * 100
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 60,
            f'{val}\n({pct:.1f}%)', ha='center', fontsize=10, fontweight='bold')

ax.set_title('Class Distribution — UCI Phishing Dataset', fontsize=14,
             fontweight='bold', pad=15)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.set_ylim(0, max(counts) * 1.2)
ax.yaxis.grid(True, linestyle='--', alpha=0.4)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Total samples: {len(y)}")
print(f"Legitimate: {counts[0]} | Suspicious: {counts[1]} | Phishing: {counts[2]}")
```

```python
#  PLOT 4: Summari
fig, axes = plt.subplots(1, 4, figsize=(12, 3.5))
fig.suptitle('UCI Phishing Websites Dataset — Summary', fontsize=14,
             fontweight='bold', y=1.02)

stats = [
    ('Total Samples', f'{len(x):,}', '#1565c0'),
    ('Features', f'{x.shape[1]}', '#2e7d32'),
    ('Phishing', f'{(y==1).sum():,}\n({(y==1).mean()*100:.1f}%)', '#c62828'),
    ('Legitimate', f'{(y==-1).sum():,}\n({(y==-1).mean()*100:.1f}%)', '#2e7d32'),
]

for ax, (label, value, color) in zip(axes, stats):
    ax.set_facecolor(color)
    ax.text(0.5, 0.6, value, transform=ax.transAxes,
            ha='center', va='center', fontsize=22, fontweight='bold',
            color='white')
    ax.text(0.5, 0.18, label, transform=ax.transAxes,
            ha='center', va='center', fontsize=11, color='white', alpha=0.9)
    ax.set_xticks([]); ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(False)

plt.tight_layout()
plt.savefig('dataset_summary.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
x_trn,x_tst,y_trn,y_tst = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
scaler = joblib.load("scaler.pkl")
x_trn = scaler.transform(x_trn)
x_tst = scaler.transform(x_tst)
```

```python
y_trn_xgb = np.where(y_trn == -1, 0, y_trn)
y_tst_xgb = np.where(y_tst == -1, 0, y_tst)

xgb = XGBClassifier(n_estimators=100,
                    eval_metric='logloss',
                    random_state=42)
xgb.fit(x_trn, y_trn_xgb)
```

```python
start = time.time()
y_pred = xgb.predict(x_tst)
time1 = time.time() - start

tracemalloc.start()
_ = xgb.predict(x_tst)
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"Prediction Time  : {time1:.6f} seconds")
print(f"Memory (current) : {current_mem / 1024:.2f} KB")
print(f"Memory (peak)    : {peak_mem / 1024:.2f} KB")
```

```python
joblib.dump(xgb, "xgbModel.pkl")
print("Model saved as xgbModel.pkl")
```

```python
importances = xgb.feature_importances_

# 2. Get the real feature names from your dataframe 
feature_names = x.columns 

# 3. Combine them into a DataFrame and sort them
importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances})
importance_df = importance_df.sort_values(by='Importance', ascending=False)

# 4. Plot the top 10 using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=importance_df.head(10), hue='Feature', palette='viridis', legend=False)

plt.title('Top 10 Most Important Features in XGBoost', fontsize=14, pad=15)
plt.xlabel('Importance Score', fontsize=12)
plt.ylabel('Feature Name', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
train_accuracy = xgb.score(x_trn, y_trn_xgb)
test_accuracy  = accuracy_score(y_tst_xgb, y_pred)

print("\n========== OVERFITTING CHECK ==========")
print(f"Training Accuracy : {train_accuracy:.4f} ({train_accuracy*100:.2f}%)")
print(f"Test Accuracy     : {test_accuracy:.4f}  ({test_accuracy*100:.2f}%)")
print(f"Difference        : {abs(train_accuracy - test_accuracy):.4f}")

if abs(train_accuracy - test_accuracy) < 0.02:
    print("Status            : No Overfitting (difference < 2%)")
elif abs(train_accuracy - test_accuracy) < 0.05:
    print("Status            : Slight Overfitting (difference 2-5%)")
else:
    print("Status            : Overfitting detected (difference > 5%)")
```

```python
accuracy        = accuracy_score(y_tst_xgb, y_pred)
precision       = precision_score(y_tst_xgb, y_pred)
recall          = recall_score(y_tst_xgb, y_pred)
f1score         = f1_score(y_tst_xgb, y_pred)
confusionMatrix = confusion_matrix(y_tst_xgb, y_pred)

print("\n========== XGBOOST RESULTS ==========")
print(f"Accuracy  = {accuracy:.4f}  ({accuracy*100:.2f}%)")
print(f"Precision = {precision:.4f}  ({precision*100:.2f}%)")
print(f"Recall    = {recall:.4f}  ({recall*100:.2f}%)")
print(f"F1 Score  = {f1score:.4f}  ({f1score*100:.2f}%)")
print(f"\nPrediction Time : {time1:.6f} sec")
print(f"Peak Memory     : {peak_mem / 1024:.2f} KB")
```

```python

disp = ConfusionMatrixDisplay(confusion_matrix=confusionMatrix,
                              display_labels=[-1, 1])
disp.plot(cmap='Blues')

plt.title("Confusion Matrix")
plt.savefig('Consusion.png', dpi=150)
plt.show()

```

```python
y_prob = xgb.predict_proba(x_tst)[:, 1]
fpr, tpr, _ = roc_curve(y_tst_xgb, y_prob)
roc_auc = auc(fpr, tpr)

print(f"\nROC-AUC Score = {roc_auc:.4f}")
```

```python
plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, color='darkorange', lw=2,
         label=f"XGBoost (AUC = {roc_auc:.4f})")
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', lw=1)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve (AUC = %0.4f)" % roc_auc)
plt.legend(loc="lower right")
plt.tight_layout()
plt.savefig("roc_xgboost.png", dpi=150)
plt.show()
print("ROC curve saved as roc_xgboost.png")
```

```python
summary = pd.DataFrame({
    "Model":         ["XGBoost"],
    "Accuracy (%)":  [round(accuracy * 100, 2)],
    "Precision (%)": [round(precision * 100, 2)],
    "Recall (%)":    [round(recall * 100, 2)],
    "F1-Score (%)":  [round(f1score * 100, 2)],
    "ROC-AUC":       [round(roc_auc, 4)],
    "Pred Time (s)": [round(time1, 6)],
    "Peak Mem (KB)": [round(peak_mem / 1024, 2)]
})

print("\n========== SUMMARY TABLE ==========")
print(summary.to_string(index=False))
```
